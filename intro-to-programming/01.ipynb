## 1. Separate Letters and Numbers

Ask the user for a string that contains both letters and numbers (e.g., `"Py2th0on3"`). Write a program that separates them into two lists: one for the letters and one for the numbers. Calculate the **sum** of the numbers found and **join** the letters to form a word.

**Example**

| Input | Output |
| --- | --- |
| `"Py2th0on3"` | Word: `"Python"`, Sum: `5` |

````python
user_input = input("Please enter a string contains letters & numbers:")

letters = []
numbers = []

for char in user_input:
    if char.isdigit():
        numbers.append(int(char))
    else:
        letters.append(char)

print(f"Word: {''.join(letters)}, Sum: {sum(numbers)}")
````

## 2. Password Validator

## 3. Store Discount Calculator

A store gives discounts based on the total price:

| Total Price | Discount |
| --- | --- |
| Under `$50` | No discount |
| `$50` to `$100` | 10% |
| Over `$100` | 20% |

If the user is a **Member**, they get an additional **5% off** the final calculated price.

**Example**

| Input | Output |
| --- | --- |
| Price: `100`, Member: `Yes` | Final Total: `$85.5` |

> Note: `100 - 10% = 90`; then `90 - 5% = 85.5`

````python
price = 100
is_member = True


price_after_discount = price

if price > 100:
    price_after_discount -= price_after_discount * 20 / 100
elif price >= 50 and price <= 100:
    price_after_discount -= price_after_discount * 10 / 100


if is_member:
  price_after_discount -= price_after_discount * 5 / 100

price_after_discount
````

## 4. Common Names

**Example**

| Input | Output |
| --- | --- |
| List 1: `"Ali, Ben, Sara"`<br>List 2: `"Ben, Joy, Ali"` | Common names: `['Ali', 'Ben']` |

````python
set_one = {
  name.strip().lower()
  for name in input("Enter first list of names (comma-separated): ").split(",")
}

set_two = {
  name.strip().lower()
  for name in input("Enter first list of names (comma-separated): ").split(",")
}

[]

intersection_list = list(set_one.intersection(set_two)) # set_one & set_two (intersection)

print(f"Common names: {intersection_list}")
````

## 5. Honor Roll

Create a dictionary where keys are student names and values are lists of 3 test scores. Write a program that calculates the **average** for each student. If the average is **> 90**, print the student's name followed by `"Honor Roll."`

**Example**

| Input | Output |
| --- | --- |
| `{'Sam': [95, 90, 92]}` | `Sam: Average 92.3 - Honor Roll` |

````python
students = {
  "Sam": [95, 90, 92],
  "Ali": [20, 40, 30]
}

for student_name, scores in students.items():
  average = sum(scores) / len(scores)
  if (average > 90):
    print(f"{student_name}: Average {average:.2f} - Honor Roll")
````

## 7. Digital Root

Ask for a multi-digit number. Sum its digits. If the result is more than one digit, keep summing the digits until you get a **single-digit** result.

**Example**

| Input | Output |
| --- | --- |
| `98` | Digital Root: `8` |

> `9 + 8 = 17`, then `1 + 7 = 8`

````python
user_input = input("Please enter a multi digit number").strip();

if user_input.isdigit():
  numbers_list = [int(number) for number in user_input]
  sum_result = sum(numbers_list)
  while sum_result > 9:
    numbers_list = [int(number) for number in str(sum_result)]
    sum_result = sum(numbers_list)
  print(f"Digital root: {sum_result}")
else:
  print("Please enter a valid number")
````

## 9. Factorial Divisibility

Ask the user for a number `n`. Calculate the factorial (`n!`) using a loop, then check if the resulting factorial is **divisible** by the sum of numbers from `1` to `n`.

**Example**

| Input | Output |
| --- | --- |
| `4` | `Factorial is 24, Sum is 10. 24 is not divisible by 10.` |

````python
input_value = 4;

sum_result = 0
factorial_result = 1
for number in range(1, input_value + 1):
    sum_result += number
    factorial_result *= number

print(f"Factorial is {factorial_result}, Sum is {sum_result}. {factorial_result} is { "divisible" if factorial_result % sum_result == 0 else "not divisible "} by {sum_result}")
````

## 10. Shopping Cart

Start with a dictionary of items and prices:

```python
{'apple': 0.5, 'bread': 2.0, 'milk': 1.5}
```

Allow the user to **Add** or **Remove** items from a cart. If they add an item not in the store, print an error. Finally, print the **total bill**.

**Example**

| Input | Output |
| --- | --- |
| Add `'apple'`, Add `'milk'` | Total Bill: `$2.0` |

````python
store = {
  "apple": 0.5,
  "bread": 2.0,
  "milk": 1.5
}

cart = {}

while True:
  operation = input("Type 'Add', 'Remove' or 'Exit' for your intent").strip().lower()
  if operation not in ['add', 'remove', 'exit']:
    print("Operation not allowed ❌")
    break

  if operation == 'exit':
     print("Goodbye")
     print(f"Total bill ${sum([store.get(product_name) * product_count for product_name, product_count in cart.items()])}")
     break

  item = input(f"Enter item name {list(store.keys())}").strip().lower()
  if item not in store:
        print("Item not allowed ❌")
        break

  item_count_in_cart = cart.get(item)

  if operation == 'add':
     if item_count_in_cart == None:
        cart[item] = 1
     else:
        cart[item] +=1
  elif operation == 'remove':
     if item_count_in_cart == None:
        continue
     elif item_count_in_cart == 1:
        cart.pop(item)
     else:
        cart[item] -=1
````

## 11. Sentence Palindrome

Check if a full sentence is a **palindrome**, ignoring spaces and capitalization.

**Example**

| Input | Output |
| --- | --- |
| `"Race car"` | `True` |

````python
user_input = input("Enter a word to check palindrome: ").strip()

sanitized_user_input = user_input.lower().replace(" ", '')
if sanitized_user_input == sanitized_user_input[::-1]:
  print(f"{user_input} -> True")
else:
  print(f"{user_input} -> False")
````

## 14. Flip a Dictionary

Take a dictionary like `{'A': 1, 'B': 2, 'C': 1}` and **flip** it. If two keys have the same value, the new value should be a **list** of those keys.

**Example**

| Input | Output |
| --- | --- |
| `{'A': 1, 'B': 2, 'C': 1}` | `{1: ['A', 'C'], 2: ['B']}` |

````python
original = {'A': 1, 'B': 2, 'C': 1}

flipped = {}

for key, value in original.items():
    if value not in flipped:
        flipped[value] = []
    flipped[value].append(key)

print(flipped)
````

## 15. Vowel Indices

Create a dictionary where the keys are the **vowels** found in a user's string and the values are lists of the **indices** where those vowels appear.

**Example**

| Input | Output |
| --- | --- |
| `"banana"` | `{'a': [1, 3, 5]}` |

````python
user_input = input("Enter a string: ").lower()

result = {}

for index, char in enumerate(user_input):
    if char not in "aeiou":
        continue

    if char not in result:
        result[char] = []
    result[char].append(index)


print(result)
````
