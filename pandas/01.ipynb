## Task 2: Handling Missing Values

The dataset contains missing values in the `Salary` and `Bonus` columns.

**Requirements:**
- Fill the missing `Salary` values with the median salary of the entire company
- Fill the missing `Bonus` values with 0

```python
salary_median = df['Salary'].median()

df['Salary'] = df['Salary'].fillna(salary_median)
df['Bonus'] = df['Bonus'].fillna(0)

df
```

## Task 3: Data Cleaning

The `Employee_Name` column is in the format "FirstName LastName".

**Requirements:**
- Split this column into two new columns: `First_Name` and `Last_Name`
- Delete the original `Employee_Name` column

```python
df[['First_Name', 'Last_Name']] = df['Employee_Name'].str.split(' ', n=1, expand=True)

df.drop(columns=['Employee_Name'], inplace=True)

df.head()
```

## Task 4: Filtering

Create a new DataFrame called `Senior_IT`.

**Requirements:**
- The DataFrame should contain only employees from the 'IT' department
- These employees must have more than 8 years of experience

```python
Senior_IT = df[(df['Department'] == 'IT') & (df['Years_Experience'] > 8)].reset_index(drop=True)

Senior_IT
```

## Task 5: Performance Mapping

Create a new column called `Performance_Level` based on the `Evaluation_Score` (1-100).

**Requirements:**
- Score > 80: "High"
- Score 50-80: "Medium"
- Score < 50: "Low"

```python
df['Performance_Level'] = df['Evaluation_Score'].apply(lambda x: 'High' if x > 80 else ('Medium' if x >= 50 else 'Low'))
df
```

## Task 6: Department Statistics

Calculate statistics for each Department.

**Requirements:**
- Calculate the total salary expense for each department
- Calculate the average evaluation score for each department

```python
total_salary = df.groupby('Department')['Salary'].sum()
print("Total salary expense for each department:")
print(total_salary)

avg_score = df.groupby('Department')['Evaluation_Score'].mean()
print("\nAverage evaluation score for each department:")
print(avg_score)
```

## Task 7: Top Earners

Find the top 5 highest-paid employees in the company.

## Task 8: Salary Update

The company is giving a 10% raise to everyone in the 'Sales' department.

**Requirements:**
- Update the `Salary` column for Sales department employees only
- Apply a 10% increase to their current salary

```python
is_sales_condition = df['Department'] == 'Sales'
df.loc[is_sales_condition, 'Salary'] = df.loc[is_sales_condition, 'Salary'] * 1.10

df
```
